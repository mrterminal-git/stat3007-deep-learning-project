#### Train and Validation Curve (with Transformer and with Weather Data)

```python
import re
import matplotlib.pyplot as plt

def parse_sharpe_log(file_path):
    epochs = []
    train_sharpe = []
    val_sharpe = []
    pattern = re.compile(r"Epoch\s+(\d+)\s+\|\s+Train Sharpe:\s+([-\d.]+)\s+\|\s+Val Sharpe:\s+([-\d.]+)")
    with open(file_path, "r") as f:
        for line in f:
            match = pattern.match(line)
            if match:
                epochs.append(int(match.group(1)))
                train_sharpe.append(float(match.group(2)))
                val_sharpe.append(float(match.group(3)))
    return epochs, train_sharpe, val_sharpe

log_path = "../data/transformer_weather_train_val_results.yml"
epochs, train_sharpe, val_sharpe = parse_sharpe_log(log_path)

plt.figure(figsize=(10,6))
plt.plot(epochs, [-x for x in train_sharpe], label="Negative Train Sharpe")
plt.plot(epochs, val_sharpe, label="Validation Sharpe")
plt.xlabel("Epoch")
plt.ylabel("Sharpe Ratio")
plt.title("Train and Validation Curve (with Transformer and Weather Data)")
plt.legend()
plt.grid(True)
plt.xscale("log")
plt.show()
```
